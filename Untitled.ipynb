```python
import os
import pandas as pd

# The language files are in the current folder
folder = "."

language_codes = {
    "ar": "Arabic",
    "zh": "Chinese",
    "cs": "Czech",
    "da": "Danish",
    "nl": "Dutch",
    "en": "English",
    "eo": "Esperanto",
    "fil": "Filipino",
    "fi": "Finnish",
    "fr": "French",
    "fr-CA-u-sd-caqc": "French (CA)",
    "de": "German",
    "hi": "Hindi",
    "hu": "Hungarian",
    "it": "Italian",
    "ja": "Japanese",
    "kab": "Kabyle",
    "tlh": "Klingon",
    "ko": "Korean",
    "no": "Norwegian",
    "fa": "Persian",
    "pl": "Polish",
    "pt": "Portuguese",
    "ru": "Russian",
    "es": "Spanish",
    "sv": "Swedish",
    "th": "Thai",
    "tr": "Turkish"
}

rows = []

for code, language in language_codes.items():

    filepath = os.path.join(folder, code)

    if not os.path.isfile(filepath):
        print("Not found:", code)
        continue

    print("Reading:", code)

    with open(filepath, "r", encoding="utf-8") as f:

        for line in f:

            word = line.strip()

            if word:
                rows.append({
                    "language": language,
                    "bad_word": word
                })


# Create DataFrame
df = pd.DataFrame(rows)

# Remove duplicates
df = df.drop_duplicates()

# Save CSV
df.to_csv(
    "multilingual_bad_words.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\n==============================")
print("Dataset created!")
print("Total words:", len(df))
print("==============================")

print(df.head(20))
```

```python
import pandas as pd

df = pd.read_csv("multilingual_bad_words.csv")

bad_words = set(
    df["bad_word"]
    .str.lower()
)
```

```python
def lowercase_text(text):
    return text.lower()
```

```python
text = "F.U.C.K"

print(lowercase_text(text))
```

```python
def normalize_text(text):

    text = text.lower()

    result = ""

    for ch in text:

        if ch.isalnum():
            result += ch

    return result
```

```python
def detect_bad_word(text):

    cleaned = normalize_text(text)

    for word in bad_words:

        if word in cleaned:
            return True, word

    return False, None
```

```python
comment = input("Enter comment: ")

detected, word = detect_bad_word(comment)

if detected:
    print("BAD WORD DETECTED")
    print("Matched:", word)
else:
    print("✅ No bad word detected")
```
